# Text Summarization using a Pretrained Hugging Face Model

**Objective:** Use a pretrained Hugging Face model to generate a concise summary of a long paragraph.

**Steps covered:**
1. Install the Hugging Face `transformers` library
2. Load a pretrained text summarization model
3. Enter a paragraph (~200–300 words)
4. Generate a summary
5. Compare original word count vs. summary word count

> **Note:** This notebook downloads a pretrained model from the Hugging Face Hub, so it needs an active internet connection. It runs as-is on **Google Colab** or any local machine with internet access.

## 2. Load a pretrained text summarization model

We use `facebook/bart-large-cnn`, a BART model fine-tuned specifically for abstractive
summarization on the CNN/DailyMail news dataset. It's one of the most widely used
summarization models on the Hugging Face Hub.

```
from transformers import pipeline

# Load the pretrained summarization pipeline
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")
```

## 3. Enter a paragraph (~200–300 words)

```
paragraph = """
Artificial intelligence has rapidly transformed from a niche academic pursuit into one of
the most influential technologies of the twenty-first century. Over the past decade,
advances in machine learning, particularly deep learning, have enabled computers to
perform tasks that once seemed exclusively human, such as recognizing images, understanding
natural language, and even generating creative content like music and art. This progress has
been driven by three major factors: the availability of massive datasets, significant
increases in computational power through specialized hardware like GPUs, and breakthroughs
in neural network architectures such as transformers. Large language models, trained on
vast amounts of text from the internet, can now write essays, answer questions, translate
languages, and assist with programming tasks. Beyond language, AI systems are being deployed
in healthcare to detect diseases from medical scans, in finance to identify fraudulent
transactions, and in transportation to power self-driving vehicles. However, this rapid
growth has also raised important concerns. Questions about bias in training data, the
environmental cost of training large models, job displacement due to automation, and the
potential misuse of AI for misinformation or surveillance have become central topics of
public debate. Governments and organizations worldwide are now working to establish
regulatory frameworks that balance innovation with safety and ethical responsibility.
As AI continues to evolve, researchers are also exploring how to make these systems more
transparent, interpretable, and aligned with human values, ensuring that the benefits of
this powerful technology are distributed fairly across society while minimizing potential
harms.
"""

print(paragraph)
```

## 4. Generate a summary

```
summary_result = summarizer(
    paragraph,
    max_length=60,   # maximum length of the summary (in tokens)
    min_length=20,   # minimum length of the summary (in tokens)
    do_sample=False
)

summary_text = summary_result[0]["summary_text"]
print("SUMMARY:\n")
print(summary_text)
```

## 5. Compare original word count vs. summary word count

```
original_word_count = len(paragraph.split())
summary_word_count = len(summary_text.split())
reduction_pct = (1 - summary_word_count / original_word_count) * 100

print(f"Original paragraph word count : {original_word_count}")
print(f"Summary word count            : {summary_word_count}")
print(f"Reduction                     : {reduction_pct:.1f}%")
```

```
import pandas as pd

comparison_df = pd.DataFrame({
    "Metric": ["Word Count", "Character Count"],
    "Original": [original_word_count, len(paragraph)],
    "Summary": [summary_word_count, len(summary_text)]
})
comparison_df
```

## Summary

- **Model used:** `facebook/bart-large-cnn` (pretrained, abstractive summarization)
- The model condensed a ~250-word paragraph into a short summary while preserving the
  core meaning.
- Word count comparison confirms a significant reduction in length, demonstrating the
  model's ability to compress information without losing key content.
