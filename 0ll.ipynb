```python
import ollama

def generate_response(model_name: str, prompt: str) -> str:
    response = ollama.chat(
        model=model_name,
        messages=[{"role": "user", "content": prompt}]
    )
    return response["message"]["content"]

def run_prompt_test(model_name: str, prompts: list):
    results = {}
    for i, prompt in enumerate(prompts, start=1):
        print(f"\n{'='*60}")
        print(f"Prompt {i}: {prompt}")
        print(f"{'='*60}")
        output = generate_response(model_name, prompt)
        print(f"Response:\n{output}\n")
        results[prompt] = output
    return results

model = "llama3.2"

user_prompt = input("Enter your prompt: ")
print("\nGenerating response...\n")
answer = generate_response(model, user_prompt)
print("Model response:")
print(answer)

test_prompts = [
    "What is the capital of France?",
    "Write a short poem about the ocean.",
    "If a train travels 60 km in 1.5 hours, what is its average speed?",
    "List three benefits of exercise in bullet points.",
    "Explain the difference between AI and Machine Learning in simple terms."
]
comparison = run_prompt_test(model, test_prompts)
```
