# Text Generation with Hugging Face Transformers

**Objective:** Understand how Generative AI models can generate new text based on a given input or prompt.

**Tech Stack:** Python, Hugging Face Transformers, PyTorch, Jupyter Notebook

**What this notebook does:**
1. Loads a pretrained text-generation model (GPT-2) from Hugging Face.
2. Accepts a sentence or topic from the user.
3. Generates text based on the given input.
4. Displays the generated text.
5. Runs the program on at least 3 different inputs.

## Step 1: Import Libraries

```
import torch
from transformers import pipeline, set_seed

print("PyTorch version:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())
```

## Step 2: Load a Pretrained Text-Generation Model

We use **GPT-2**, a pretrained causal language model available on the Hugging Face Hub,
via the `transformers` `pipeline` API. This downloads the model the first time it runs
(internet connection required) and caches it locally afterwards.


```
device = 0 if torch.cuda.is_available() else -1

generator = pipeline("text-generation", model="gpt2", device=device)
set_seed(42)  # for reproducibility

print("Model loaded successfully!")
```

## Step 3: Define a Function to Generate Text

This function:
- Accepts a sentence or topic (prompt) from the user.
- Generates new text continuing from that prompt.
- Returns/displays the generated text.


```
def generate_text(prompt, max_length=60, num_return_sequences=1):
    """
    Generate text based on a given input prompt using a pretrained
    Hugging Face text-generation model.

    Parameters:
        prompt (str): The input sentence or topic to generate text from.
        max_length (int): Maximum length (in tokens) of the generated text.
        num_return_sequences (int): Number of generated text variations to return.

    Returns:
        list[str]: A list of generated text strings.
    """
    results = generator(
        prompt,
        max_length=max_length,
        num_return_sequences=num_return_sequences,
        truncation=True,
        pad_token_id=generator.tokenizer.eos_token_id
    )
    return [r["generated_text"] for r in results]


def display_generated_text(prompt, generated_texts):
    """Nicely display the prompt and the generated text(s)."""
    print("=" * 80)
    print(f"INPUT PROMPT:\n{prompt}")
    print("-" * 80)
    for i, text in enumerate(generated_texts, start=1):
        print(f"GENERATED TEXT (Option {i}):\n{text}")
        print("-" * 80)
```

## Step 4: Accept Input From the User (Interactive)

Run the cell below to type your own sentence or topic and generate text from it interactively.
(If running in an environment without interactive input, skip to Step 5 for preset examples.)


```
# Interactive input — run this cell and type a prompt when asked
try:
    user_prompt = input("Enter a sentence or topic to generate text from: ")
    if user_prompt.strip():
        generated = generate_text(user_prompt)
        display_generated_text(user_prompt, generated)
    else:
        print("No input provided.")
except Exception as e:
    print("Interactive input not available in this environment:", e)
```

## Step 5: Test the Program With Multiple Inputs

Here we try the text generator with **three different inputs** to see how it performs
on different sentences/topics.


```
test_prompts = [
    "Artificial intelligence is transforming the way we",
    "Once upon a time in a small village,",
    "The future of renewable energy depends on"
]

for prompt in test_prompts:
    generated = generate_text(prompt, max_length=60)
    display_generated_text(prompt, generated)
```

## Step 6: Try Your Own Additional Inputs

Feel free to add more prompts to the list below and re-run to explore how the model
behaves with different topics, tones, and lengths.


```
more_prompts = [
    "My favorite hobby is",
    # Add your own prompts here, e.g. "Climate change will",
]

for prompt in more_prompts:
    generated = generate_text(prompt, max_length=50)
    display_generated_text(prompt, generated)
```

## Observations

- The model (GPT-2) generates text by predicting the most likely next word/token,
  one step at a time, based on the input prompt and everything generated so far.
- Because generation involves some randomness (sampling), the same prompt can produce
  different outputs on different runs (unless a fixed seed like `set_seed(42)` is used).
- Longer `max_length` values produce more text, but quality/coherence can drift the
  further the model generates from the original prompt.
- Try changing `max_length`, `num_return_sequences`, or the model name (e.g. `"gpt2-medium"`,
  `"distilgpt2"`) to see how results change.
